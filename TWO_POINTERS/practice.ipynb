## Two Sum

```python
arr = [2, 5, 7, 12, 13, 15, 16]#when 16 ka minimum(2) ke sath hi sum jyada aa rha , to baki ke sath to aur jyada aayega , so 16 can not be part of two sum pair, we can discard 16 
target = 12

def two_sum(arr, target):
    left = 0
    right = len(arr) - 1

    while left<right:
        current_sum = arr[left] + arr[right]
        if current_sum == target:
            print(f"Pair found: {arr[left]}, {arr[right]}")
            return arr[left], arr[right]
        elif current_sum>target:
            right-=1
            print(current_sum,f'current sum is less than {target} so taking bigger number by moving left pointer to right')
        else:
            left+=1
            print(current_sum, f'current sum is greater than {target} so taking smaller number by moving right pointer to left')
    return None

result = two_sum(arr, target)
```

## Reverse an array

```python
arr = [10, 20, 30, 40, 50]

def reverse_array(arr):
    left = 0
    right = len(arr) - 1

    while left<right:
        arr[left], arr[right] = arr[right], arr[left]
        left+=1
        right-=1
    return arr

print(reverse_array(arr))
```

## Reverse an string

```python
s = 'abcdef'
strlen = len(s)
ns = ''

for i in range(strlen):
    ix = strlen - 1 - i
    ns = ns + s[ix]

print(ns)
```

##Check whether a string is a palindrome

```python
s = 'racecar'

left = 0
right = len(s)-1

while left<right:
    if s[left]!=s[right]:
        print('not palindrmome')
        break
    left+=1
    right-=1
else:
    print('palindrome')
```

## Remove duplicates from a sorted array

```python
def remove_duplicates(arr):

    if not arr:
        return []

    slow = 0  # Tu pehle unique element  par reh.

    for fast in range(1, len(arr)):  # Main ek-ek element check karunga.

        if arr[fast] != arr[slow]:  # Agar mujhe naya unique element mila...

            slow += 1  # Tu next available position par ja.

            arr[slow] = arr[fast]  # Aur mera unique element wahan rakh de.

    return arr[:slow + 1]  # jitna sequence tune banaya hai, usko return kar.
```

## REMOVE ELEMENT

```python
nums =  [0, 1, 3, 0, 2, 2, 4, 2]
val = 2

left = 0 # tu first element per reh
for right in range(len(nums)): # mai ek ek krke saare element pr jaunga
    if nums[right]!=val: #vo element pick kruga jo target ke brabar nhi hai
        nums[left] = nums[right] # tu apne place pr rakh kar
        left+=1 # next position pr chle jana

nums[:left+1]
```

##Move Zeroes

```python
nums =  [0, 1, 0, 3, 12] #non zero one by one pick krke using right pointer, left side rkhte ja rhe using left right pointer
left = 0
for right in range(len(nums)):
    if nums[right]!=0:
        nums[left] = nums[right]
        left+=1
nums[left:] = [0]*(len(nums)-left)
```
